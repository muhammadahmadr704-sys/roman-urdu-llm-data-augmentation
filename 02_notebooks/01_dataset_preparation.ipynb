```python
import pandas as pd

# ============================================================
# Load the final prepared RUHSOLD dataset
# ============================================================

train_df = pd.read_csv(
    "RUHSOLD_train.tsv",
    sep="\t",
    names=["tweet", "label"]
)

val_df = pd.read_csv(
    "RUHSOLD_validation.tsv",
    sep="\t",
    names=["tweet", "label"]
)

test_df = pd.read_csv(
    "RUHSOLD_test.tsv",
    sep="\t",
    names=["tweet", "label"]
)

print(f"Training samples   : {len(train_df)}")
print(f"Validation samples : {len(val_df)}")
print(f"Testing samples    : {len(test_df)}")
```

```python
train_set = set(train_df["tweet"].astype(str))
val_set = set(val_df["tweet"].astype(str))
test_set = set(test_df["tweet"].astype(str))

print("Train ∩ Validation :", len(train_set & val_set))
print("Train ∩ Test       :", len(train_set & test_set))
print("Validation ∩ Test  :", len(val_set & test_set))
```
